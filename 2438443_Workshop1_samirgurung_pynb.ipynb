```python
# 1. Fibonacci series up to 10 terms
a = 0
b = 1
print("Fibonacci Series:")
for i in range(10):
    print(a, end=" ")
    temp = a
    a = b
    b = temp + b
print()


```

```python
# 2. Factorial of a given number
num = int(input("Enter a number to find factorial: "))
result = 1
i = 1
while i <= num:
    result = result * i
    i = i + 1
print("Factorial of", num, "is", result)

```

```python

# 3. Right angle triangle
rows = 5
for i in range(1, rows + 1):
    stars = ""
    for j in range(i):
        stars = stars + "* "
    print(stars)
```

```python


# 4. Simple calculator
def add(x, y):
    return x + y

def subtract(x, y):
    return x - y

def multiply(x, y):
    return x * y

def divide(x, y):
    if y == 0:
        return "Cannot divide by zero"
    return x / y

a = float(input("Enter first number: "))
b = float(input("Enter second number: "))

print("Addition:", add(a, b))
print("Subtraction:", subtract(a, b))
print("Multiplication:", multiply(a, b))
print("Division:", divide(a, b))
```

```python

# 5. Remove the first duplicate element of a list
my_list = [1, 2, 3, 2, 4, 5, 3, 6]
print("Original list:", my_list)

seen = []
for item in my_list:
    if item in seen:
        my_list.remove(item)
        print("Removed first duplicate:", item)
        break
    else:
        seen.append(item)

print("List after removal:", my_list)

```

```python

# 6. Extract mentioned keys from a dictionary
sample_dict = {"name": "Anderson", "age": 35, "salary": 5000, "city": "London"}
keys = ["name", "salary"]

new_dict = {}
for k in keys:
    if k in sample_dict:
        new_dict[k] = sample_dict[k]

print("New dictionary:", new_dict)
```

```python


# 7. Convert two lists into a dictionary
list_one = ['one', 'two', 'three']
list_two = [1, 2, 3]

new_dict = {}
for i in range(len(list_one)):
    new_dict[list_one[i]] = list_two[i]

print("Dictionary:", new_dict)
```
